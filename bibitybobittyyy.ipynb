```python
all_ml_data = {}
for stock in ticks:
    stock_features = pd.DataFrame(index=prices.index)
    stock_features["return_1d"] = returns[stock]
    stock_features["return_5d"] = prices[stock].pct_change(5)
    stock_features["return_20d"] = prices[stock].pct_change(20)
    # Moving averages
    stock_features["ma_20"] = prices[stock].rolling(20).mean()
    stock_features["ma_50"] = prices[stock].rolling(50).mean()
    # Stock risk
    stock_features["volatility_20d"] = returns[stock].rolling(20).std()
    # Market features
    stock_features["market_return"] = returns.mean(axis=1)
    stock_features["market_volatility"] = returns.std(axis=1)
    # Target = next day's return
    stock_features["target"] = returns[stock].shift(-1)




    stock_features = stock_features.replace(
        [np.inf, -np.inf], np.nan
    ).dropna()
    all_ml_data[stock] = stock_features
print("Stocks prepared:", len(all_ml_data))
```

```python
models = {}
predicted_returns = {}
evaluation = {}




for stock in ticks:

    data = all_ml_data[stock]
    split = int(len(data) * 0.8)
    train = data.iloc[:split]
    test = data.iloc[split:]
    X_train = train.drop(columns="target")
    y_train = train["target"]
    X_test = test.drop(columns="target")
    y_test = test["target"]
    model = XGBRegressor(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.0025,
        random_state=45
    )
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    models[stock] = model
    predicted_returns[stock] = predictions[-1]




    evaluation[stock] = {
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2": r2_score(y_test, predictions)
    }
print("Models trained:", len(models))
```

```python
from sklearn.model_selection import TimeSeriesSplit, RandomizedSearchCV
from xgboost import XGBRegressor

tscv = TimeSeriesSplit(n_splits=3)
param_grid = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [2, 3, 4, 5],
    "learning_rate": [0.01, 0.03, 0.05, 0.1],
    "subsample": [0.7, 0.8, 1.0],
    "colsample_bytree": [0.7, 0.8, 1.0]
}
print("Hyperparameter tuning setup complete.")
```

```python
stock = "RELIANCE.NS"
data = all_ml_data[stock]
split = int(len(data) * 0.8)

train = data.iloc[:split]
test = data.iloc[split:]

X_train = train.drop(columns="target")
y_train = train["target"]

X_test = test.drop(columns="target")
y_test = test["target"]

base_model = XGBRegressor(
    objective="reg:squarederror",
    random_state=42
)
search = RandomizedSearchCV(
    estimator=base_model,
    param_distributions=param_grid,
    n_iter=20,
    scoring="r2",
    cv=tscv,
    random_state=42,
    n_jobs=-1
)

search.fit(X_train, y_train)

print("Best parameters:")
print(search.best_params_)

print("\nBest CV R²:")
print(search.best_score_)
```

```python
tuned_models = {}
tuned_predictions = {}
tuned_evaluation = {}
best_params = {}

for stock in ticks:

    print(f"Tuning {stock}...")
    data = all_ml_data[stock]
    split = int(len(data) * 0.8)
    train = data.iloc[:split]
    test = data.iloc[split:]
    X_train = train.drop(columns="target")
    y_train = train["target"]
    X_test = test.drop(columns="target")
    y_test = test["target"]
    base_model = XGBRegressor(
        objective="reg:squarederror",
        random_state=42
    )
    search = RandomizedSearchCV(
        estimator=base_model,
        param_distributions=param_grid,
        n_iter=20,
        scoring="r2",
        cv=tscv,
        random_state=42,
        n_jobs=-1
    )
    search.fit(X_train, y_train)
    best_model = search.best_estimator_
    predictions = best_model.predict(X_test)
    tuned_models[stock] = best_model
    tuned_predictions[stock] = predictions[-1]
    best_params[stock] = search.best_params_
    tuned_evaluation[stock] = {
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2": r2_score(y_test, predictions)
    }
print("\nTuning completed for:", len(tuned_models), "stocks")
```

```python
tuned_evaluation_df = pd.DataFrame(tuned_evaluation).T
print(tuned_evaluation_df)
```
